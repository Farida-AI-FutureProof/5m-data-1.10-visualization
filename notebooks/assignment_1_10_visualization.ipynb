```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

### Question 1

How do you create a 2x2 subplot grid in matplotlib and select the first subplot?

```python
fig, axes = plt.subplots(2, 2)  # 2x2 grid
ax = axes[0, 0]                 # first subplot (top-left)
plt.show()
```

### Question 2

How to plot a line and set the color to red and style to dash in a matplotlib plot?

```python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y, color="red", linestyle="--")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Red dashed line example")
plt.show()
```

### Question 3

How to plot a histogram with 30 bins for `data` in matplotlib?

```python
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of data")
plt.show()
```

### Question 4

How can you set the x-axis and y-axis labels in a matplotlib plot?

```python
plt.plot([1, 2, 3], [1, 4, 9])
plt.xlabel("X axis label")
plt.ylabel("Y axis label")
plt.title("Example Plot")
plt.show()
```

### Question 5

How do you create a bar plot in seaborn using the `tips` dataset to show the average tip amount per day?

```python
tips = sns.load_dataset("tips")

sns.barplot(data=tips, x="day", y="tip")  # default estimator is mean
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.title("Average Tip Amount per Day")
plt.show()
```

### Question 6

How to create a box plot for `total_bill` categorized by day in the `tips` dataset using seaborn?

```python
tips = sns.load_dataset("tips")

sns.boxplot(data=tips, x="day", y="total_bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Total Bill by Day")
plt.show()
```
